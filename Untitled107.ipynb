```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from statsmodels.tsa.seasonal import STL

# 1. Load your dataset and filter for the year 1984 starting from 1984-09-07
stock_data = pd.read_csv(r"C:\Users\ashut\Downloads\aapl_us_2025 (1).csv")
stock_data['Date'] = pd.to_datetime(stock_data['Date'])
stock_data.set_index('Date', inplace=True)
filtered_data = stock_data.loc['1984-09-07':'1984-12-31']['Close'].dropna()

# 2. Apply STL Decomposition (adjusted period for the available months in 1984)
stl = STL(filtered_data, period=10, seasonal=3)
result = stl.fit()

# 3. Define date formatting rules for Year-Month (%Y-%m) across different months of the same year
date_locator = mdates.MonthLocator()  # Show every month
date_formatter = mdates.DateFormatter('%Y-%m')  # Year-Month format

# 4. Plot all components in a 4-subplot vertical layout
plt.figure(figsize=(10, 10))

# Panel 1: Original Time Series
plt.subplot(4, 1, 1)
plt.plot(result.observed.index, result.observed.values, label='Original', color='blue')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.title('STL Decomposition: Same Year (1984) with Different Months')
plt.gca().xaxis.set_major_locator(date_locator)
plt.gca().xaxis.set_major_formatter(date_formatter)
plt.xticks(rotation=30)

# Panel 2: Trend Component
plt.subplot(4, 1, 2)
plt.plot(result.trend.index, result.trend.values, label='Trend', color='red')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.gca().xaxis.set_major_locator(date_locator)
plt.gca().xaxis.set_major_formatter(date_formatter)
plt.xticks(rotation=30)

# Panel 3: Seasonal Component
plt.subplot(4, 1, 3)
plt.plot(result.seasonal.index, result.seasonal.values, label='Seasonal', color='green')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.gca().xaxis.set_major_locator(date_locator)
plt.gca().xaxis.set_major_formatter(date_formatter)
plt.xticks(rotation=30)

# Panel 4: Residual (Noise) Component
plt.subplot(4, 1, 4)
plt.plot(result.resid.index, result.resid.values, label='Residual', color='orange')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.gca().xaxis.set_major_locator(date_locator)
plt.gca().xaxis.set_major_formatter(date_formatter)
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
